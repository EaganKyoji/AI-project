# Setup Environment
Setup some library and dataset in df variable

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import random as rn

df = pd.read_csv('/kaggle/input/datasets/ucupsedaya/gojek-app-reviews-bahasa-indonesia/GojekAppReviewV4.0.0-V4.9.3_Cleaned.csv')
```

```python
#=======================================================
# 10 CONTOH SAMPLE DATASET 
#=======================================================
print("="*20, "Contoh 10 Baris Dataset", "="*20)
display(df[['content', 'score']].head(30).style.hide(axis='index'))
```

```python
#=======================================================
# PENGAMBILAN SAMPLE DATASET 
#=======================================================
review_A = df['content'][21]
review_B = df['content'][11]
review_C = df['content'][16]
print(review_A, "\n")
print(review_B, "\n")
print(review_C, "\n")
```

```python
#=======================================================
# PEMBERSIHAN DATASET
#=======================================================
def bersihkan(teks):
    return teks.lower().split()

kata_A = bersihkan(review_A)
kata_B = bersihkan(review_B)
kata_C = bersihkan(review_C)

print("="*20, "Contoh 10 Baris Dataset", "="*20)
print("Kata A: ",kata_A)
print("Kata B: ",kata_B)
print("Kata C: ",kata_C)

#=======================================================
# PEMBENTUKAN KATA UNIK
#=======================================================
vocab = sorted(set(kata_A + kata_B + kata_C))
print("\n", "="*20, "Vocab dari setiap kata", "="*20)
print(vocab)
print(len(vocab), "kata unik")
```

# Vektorisasi

```python
#=======================================================
# Vektorisasi
#=======================================================
def vektorisasi(tiapKata, vocab):
    posisi_kata = {kata: i for i, kata in enumerate(vocab)}
    vektor = np.zeros(len(vocab))
    for kata in tiapKata:
        if kata in vocab:
            vektor[posisi_kata[kata]] +=1
    return vektor

vA = vektorisasi(kata_A, vocab)
vB = vektorisasi(kata_B, vocab)
vC = vektorisasi(kata_C, vocab)

print("="*20, "Hasil Vektorisasi", "="*20)
print("Kata A: ",vA)
print("Kata B: ",vB)
print("Kata C: ",vC)
```

# Implementasi Matematis

```python
#=======================================================
# RUMUS FUNGSI
#=======================================================
def dot_product(v1, v2):
    return v1 @ v2

def magnitude(v):
    return np.sqrt(np.sum(v**2))

def cosine_similarity(v1, v2):
    return dot_product(v1, v2)/(magnitude(v1) * magnitude(v2))

def sudut(hasil):
    return np.degrees(np.arccos(hasil))
#=======================================================
# PEMBUKTIAN
#=======================================================
print("="*20, "Hasil akhir", "="*20)
print("A vs B (Ulasan komplain)  ", cosine_similarity(vA, vB), sudut(cosine_similarity(vA, vB)))
print("A vs C (Ulasan Berlawanan)", cosine_similarity(vA, vC), sudut(cosine_similarity(vA, vC)))
print("B vs C (Ulasan Berlawanan)", cosine_similarity(vB, vC), sudut(cosine_similarity(vB, vC)))
```

# Ulasan Acak

```python
# Ambil Variabel dengan index acak
idx = np.random.randint(0, len(df)-1, size=2)
ul = df['content'].iloc[idx].tolist()
rt = df['score'].iloc[idx].tolist()

#Bikin dataframe-nya
ulrt = pd.DataFrame({
    'Konten' : ul,
    'Skor' : rt
})
display(ulrt.style.hide(axis="index"))
```

```python
# Bersihkan dulu kontennya
ul_bersih = [bersihkan(teks) for teks in ul]
vcb_unik = []

print("="*20, "Hasil Pembersihan", "="*20)
for i in range(len(ul)):
    print(f"Ulasan {i+1}: ", ul_bersih[i])
    vcb_unik = sorted(set(ul_bersih[i]+vcb_unik))

print("\n", "="*20, "Kosa Kata Unik", "="*20)
print(f"Ada {len(vcb_unik)} kosa kata unik")
print(vcb_unik)
```

```python
# Vektorisasi
vktr = [vektorisasi(ul_bersih[i], vcb_unik) for i in range(len(ul_bersih))]

print("="*20, "Hasil Vektorisasi", "="*20)
for i in range(len(vktr)):
    print(f"Ulasan {i+1}: ", vktr[i])
```

```python
# Hasil Akhir
print("="*20, "Hasil akhir", "="*20)
csm = cosine_similarity(vktr[0], vktr[1])
print("Ulasan 1 vs Ulasan 2: ", csm, sudut(csm))
```

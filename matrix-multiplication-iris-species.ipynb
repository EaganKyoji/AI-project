```python
#=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-
# CEK TRILOGI STATISTIKA
#=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-
numeric_cols = df.select_dtypes(include='number').columns.drop('Id')

means = df[numeric_cols].mean()
medians = df[numeric_cols].median()
modes = df[numeric_cols].mode().iloc[0]

x = np.arange(len(numeric_cols))
width = 0.25

plt.figure(figsize=(12, 6)) #atur ukuran

bar1 = plt.bar(x -width, means, width, label="Means", color='#6A1B9A', zorder=2) #bikin tiap bar terus disimpen dalam variabel bar
bar2 = plt.bar(x, medians, width, label="Medians", color='#8E44AD', zorder=2)
bar3 = plt.bar(x + width, modes, width, label="Modes", color='#BA68C8', zorder=2)

plt.bar_label(bar1, fmt='%.2f', padding=3, fontsize=12) #bikin label di tiap bar
plt.bar_label(bar2, fmt='%.2f', padding=3, fontsize=12)
plt.bar_label(bar3, fmt='%.2f', padding=3, fontsize=12)

plt.ylabel('Nilai')
plt.xlabel('Fitur')
plt.title('TRILOGI STATISTIKA')

plt.xticks(x, numeric_cols)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5, zorder=0)

plt.show()


```

# Transforming Data into Matrix

```python
numMatrix = df.iloc[:, 1:5] 
numMatrix = numMatrix.to_numpy()
randomMatrix = np.array([[2, 3, 4, 5], [3, 4, 5, 6], [2, 3, 1, 2], [4, 5, 6, 7]])


print(numMatrix)
print(randomMatrix)
print(randomMatrix.shape)
```

# Matrix Operation

```python
def matmul_manual(A, B):
    baris_A, kolom_A = A.shape
    baris_B, kolom_B = B.shape
    assert kolom_A == baris_B, "Ukuran tidak cocok"
    
    hasil = np.zeros((baris_A, kolom_B))
    for i in range(baris_A):
        for j in range(kolom_B):
            total = 0
            for k in range(kolom_A):
                total += A[i][k] * B[k][j]
            hasil[i][j] = total
    return hasil
```

```python
print(np.allclose(matmul_manual(numMatrix, randomMatrix), numMatrix @ randomMatrix))
```
